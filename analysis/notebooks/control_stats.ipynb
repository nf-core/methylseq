```python
def output_control_stats(methobj, rep_list:list, run_id:str, base_data_directory:str, spike_in_controls:list = ['phage_lambda', 'plasmid_puc19c']) -> None:
    import os
    import pandas as pd

    print('\n')
    print('Percentage methylation in spike-in controls.')

    control_df_fn = base_data_directory+'/'+run_id+'/control_df.csv'

    if os.path.isfile(control_df_fn):
        control_df = pd.read_csv(control_df_fn)
    else:
        control_df = create_control_stats_table(methobj, rep_list, spike_in_controls)
        control_df.to_csv(control_df_fn)

    print(control_df)
```

```python
def create_control_stats_table(methobj, spike_in_controls:list = ['phage_lambda', 'plasmid_puc19c']) -> pd.DataFrame:
	chr_index = 0
	coverage_index = 4
	numCs_index = 5
	pct_methylatedL = []
	rep_list = list(rpy2.robjects.r.getSampleID(methobj))
	for i,this_rep in enumerate(rep_list):
		pct_methylatedL_aux = []
		for this_control in spike_in_controls:
			array_selection = np.array(methobj[i][chr_index]) == this_control
			if np.any(array_selection):
				pct_methylated = 100*np.array(methobj[i][numCs_index])[array_selection] / np.array(methobj[i][coverage_index])[array_selection]
				pct_methylated = np.mean(pct_methylated)
				pct_methylatedL_aux.append(pct_methylated)
			else:
				pct_methylatedL_aux.append(0)
		pct_methylatedL.append(pct_methylatedL_aux)
	return pd.DataFrame(pct_methylatedL, index=rep_list, columns=spike_in_controls)
```
