```python
%%R
input_string <- "[[id:healthy_1, group:healthy], [id:healthy_2, group:healthy], [id:disease_1, group:disease], [id:disease_2, group:disease]]"

# Extracting id values
id_values <- regmatches(input_string, gregexpr("(?<=id:)[^,\\]]+", input_string, perl = TRUE))[[1]]

# Extracting group values
group_values <- regmatches(input_string, gregexpr("(?<=group:)[^,\\]]+", input_string, perl = TRUE))[[1]]
group_values
```

```python
%%R
# Your input vector
input_vector <- c("healthy", "healthy", "disease", "disease")

# Find unique values and sort them alphabetically
unique_values <- sort(unique(input_vector))

# Assign numbers based on alphabetical rank
converted_vector <- match(input_vector, unique_values) - 1

# Display the converted vector
converted_vector
```

```python
%%R
char2intVec <- function(charVec){
	# Find unique values and sort them alphabetically
	unique_values = sort(unique(charVec))

	# Assign numbers based on alphabetical rank
	converted_vector = match(charVec, unique_values) - 1

	return(converted_vector)
}
```
